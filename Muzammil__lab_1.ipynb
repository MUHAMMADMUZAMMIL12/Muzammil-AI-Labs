```python
print("=== FLIGHT FINDER ===")

destinations = {
    "istanbul": 125000,
    "maldives": 180000,
    "kuala lumpur": 100000
}

destination = input("Where would you like to fly?: ").strip().lower()

try:
    max_budget = int(input("What is your maximum spend?: "))

    price = destinations.get(destination)

    if price is None:
        print("Agent: Sorry, we do not fly to that destination.")
    elif max_budget < price:
        print("Agent: That flight exceeds your budget limit.")
        print("Agent: Try searching for a more budget-friendly route.")
    else:
        print(
            "Agent: Success! You can afford a ticket to",
            destination.title()
        )

except ValueError:
    print("Agent: Please enter your budget as a number, e.g. 150000.")
```

```python
!pip install -U google-genai

from google import genai

client = genai.Client(api_key="YOUR_API_KEY")

MODEL = "gemini-2.5-flash"

# FUUAST Subject-to-Teacher Mapping Dictionary
teachers = {
    "Software Engineering": "Sir Kashif",
    "DBMS": "Miss Uzma",
    "Data Structures and Algorithms": "Dr. Shazia",
    "Multivariable Calculus": "Miss Humaira"
}


# Academic Assistant Function
def academic_assistant(user_input):
    prompt = f"""
You are an official information assistant specializing in details regarding the Federal Urdu University of Arts, Science and Technology (FUUAST).

Provide clear, accurate, and concise responses to student questions. If the requested information is not available in your reference data or online, state directly that you do not have that information.

Refer to the following subject-to-teacher mapping to answer queries about course instructors: {teachers}

User query: {user_input}
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


# AI Job Search Agent Function
def job_search_agent(user_input):
    prompt = f"""
You are a professional career counselor and AI Job Search Agent.

Provide a well-structured job search strategy with clear headings:

1. Target Roles & Industry Overview
2. Where to Apply (specific job boards, platforms, and company career portals)
3. Key Skills & Resume Keywords
4. Networking & Outreach Strategy
5. Interview Preparation & Tips

Make the response detailed, actionable, and easy for job seekers or recent graduates to follow.

User request: {user_input}

Job Search Strategy:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


# Project / System Execution Loop
print("--- AI JOB SEARCH & FUUAST ACADEMIC ASSISTANT AGENT STARTED ---")
print("(Type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye")
        break

    try:
        # Academic Assistant Call
        academic_reply = academic_assistant(user)
        print("\n[FUUAST Academic Agent]:\n", academic_reply, "\n")

        # AI Job Search Agent Call
        job_reply = job_search_agent(user)
        print("\n[AI Job Search Agent]:\n", job_reply, "\n")

    except Exception as e:
        print("Error:", e)
```
